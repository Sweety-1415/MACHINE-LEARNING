ONE TO ONE VANILLA RNN 

```python
import torch 
import torch.nn as nn
```

```python
text="what","is","your","name"
chars=sorted(list(set(text)))
print(chars)
```

```python
char_to_index={ch:i for i,ch in enumerate(chars)}
index_to_char={i:ch for ch,i in char_to_index.items()}  #for prediction convert numbers to letters
print(char_to_index)
print(index_to_char)
```

```python
encode=[char_to_index[c] for c in text] #convert text into numbers
print(text,encode)
```

```python
x=torch.tensor(encode[:-1]).unsqueeze(0) #removes last elements,unsqueeze add extra dimension(4)-->(1,4),the extra dimension is batch 1-sequence,4-time steps
y=torch.tensor(encode[1:]) #remove first element
print(x,y)
```

```python
class Simplevrnn(nn.Module):
    def __init__(self):
        super().__init__()
        self.embedding=nn.Embedding(4,8)  # 4-unique characters and 8 dimensional vector h

        self.rnn=nn.RNN(8,16,batch_first=True)
        self.fc=nn.Linear(16,4)

    def forward(self,x):
        x=self.embedding(x)
        output,hidden=self.rnn(x)  #Each time step has a hidden state of size 16.
        output=self.fc(output)   #Each time step predicts one of the unique characters.
        return output
```

```python
model=Simplevrnn()
l=nn.CrossEntropyLoss()
opti=torch.optim.Adam(model.parameters(),lr=0.01)

for epoch in range(500):
    opti.zero_grad()
    output=model(x)
    loss=l(output.view(-1,4),y)
    loss.backward()
    opti.step()

    print(loss)

print("Training Complete")
```

```python
model.eval()
input='what'
input_index=char_to_index[input]
x=torch.tensor([[input_index]]) #(1,1) -->1-batch,1-step
with torch.no_grad():
    output=model(x)
    predict=output.argmax(dim=2)
pred_char=index_to_char[predict.item()]

print("Input:",input)
print("Predicted:",pred_char)

print(input,pred_char)
```
